```python
# 示例代码：构建一个简单的图模型，并找到两点间的最短路径
import networkx as nx
import matplotlib.pyplot as plt

# 创建图模型
G = nx.Graph()

# 添加节点
G.add_node(1)
G.add_node(2)
G.add_node(3)

# 添加边
G.add_edge(1, 2)
G.add_edge(2, 3)

# 计算最短路径
path = nx.shortest_path(G, source=1, target=3)

print(f"The shortest path between node 1 and node 3 is: {path}")

# 绘制图
nx.draw(G, with_labels=True)
plt.show()
```

### Ejemplo de grafo con aristas ponderadas

Para agregar pesos a las aristas, puedes especificar el atributo `weight` al añadir las aristas. Luego, al calcular la ruta más corta, puedes indicar que `networkx` debe usar este atributo `weight`.

```python
# Crear un nuevo grafo dirigido para demostrar los pesos
import networkx as nx
import matplotlib.pyplot as plt

G_weighted = nx.DiGraph() # Usamos un grafo dirigido para mayor claridad en este ejemplo

# Añadir aristas con pesos
G_weighted.add_edge('A', 'B', weight=1)
G_weighted.add_edge('A', 'C', weight=4)
G_weighted.add_edge('B', 'C', weight=2)
G_weighted.add_edge('B', 'D', weight=5)
G_weighted.add_edge('C', 'D', weight=1)

# Calcular el camino más corto entre 'A' y 'D' considerando los pesos
# Se usa el algoritmo de Dijkstra por defecto cuando se especifica 'weight'
shortest_path_weighted = nx.shortest_path(G_weighted, source='A', target='D', weight='weight')
shortest_path_length = nx.shortest_path_length(G_weighted, source='A', target='D', weight='weight')

print(f"El camino más corto entre 'A' y 'D' (ponderado) es: {shortest_path_weighted}")
print(f"La longitud total del camino ponderado es: {shortest_path_length}")

# Visualizar el grafo con pesos en las aristas
pos = nx.spring_layout(G_weighted) # Posiciones para todos los nodos

nx.draw(G_weighted, pos, with_labels=True, node_color='lightblue', node_size=2000, font_size=10)
labels = nx.get_edge_attributes(G_weighted, 'weight')
nx.draw_networkx_edge_labels(G_weighted, pos, edge_labels=labels) # Corregido a nx.draw_networkx_edge_labels
plt.title("Grafo con Aristas Ponderadas")
plt.show()
```

### Implementación del Algoritmo de Dijkstra en Python

Aquí se presenta una implementación del algoritmo de Dijkstra utilizando `heapq` para manejar la cola de prioridad, lo que lo hace eficiente para encontrar el camino más corto en grafos ponderados.

```python
import heapq

def dijkstra(graph, start_node):
    # Inicializar distancias con infinito y predecesores como None
    distances = {node: float('infinity') for node in graph}
    predecessors = {node: None for node in graph}
    distances[start_node] = 0

    # Cola de prioridad para almacenar (distancia, nodo)
    priority_queue = [(0, start_node)] # (distancia, nodo)

    while priority_queue:
        current_distance, current_node = heapq.heappop(priority_queue)

        # Si ya hemos encontrado un camino más corto a este nodo, lo ignoramos
        if current_distance > distances[current_node]:
            continue

        # Explorar vecinos del nodo actual
        for neighbor, weight in graph[current_node].items():
            distance = current_distance + weight

            # Si se encuentra un camino más corto al vecino
            if distance < distances[neighbor]:
                distances[neighbor] = distance
                predecessors[neighbor] = current_node
                heapq.heappush(priority_queue, (distance, neighbor))

    return distances, predecessors

def reconstruct_path(predecessors, start_node, end_node):
    path = []
    current = end_node
    while current is not None:
        path.insert(0, current)
        current = predecessors[current]
    if path[0] == start_node:
        return path
    return [] # No hay camino

# Adaptar el grafo G_weighted de networkx a un formato de diccionario para Dijkstra
# networkx usa una estructura de adyacencia diferente, la convertimos para el algoritmo

# Ejemplo de cómo se vería un grafo para la implementación de Dijkstra:
# { 'A': {'B': 1, 'C': 4},
#   'B': {'C': 2, 'D': 5},
#   'C': {'D': 1},
#   'D': {} }

dijkstra_graph = {
    node: {neighbor: data['weight'] for neighbor, data in G_weighted[node].items()}
    for node in G_weighted.nodes()
}

start_node_dijkstra = 'A'
end_node_dijkstra = 'D'

# Ejecutar Dijkstra
distances, predecessors = dijkstra(dijkstra_graph, start_node_dijkstra)

# Reconstruir el camino
path_dijkstra = reconstruct_path(predecessors, start_node_dijkstra, end_node_dijkstra)

print(f"Distancias desde {start_node_dijkstra}: {distances}")
print(f"Predecesores: {predecessors}")
print(f"Camino más corto de {start_node_dijkstra} a {end_node_dijkstra}: {path_dijkstra}")
print(f"Longitud del camino: {distances[end_node_dijkstra]}")
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# 1. Configuración del Grafo Dinámico
G = nx.DiGraph() # Usamos DiGraph para representar el flujo unidireccional de las calles

# Definimos las intersecciones (Nodos)
# Ejemplo: Intersecciones de la zona de estudio en Barrancabermeja
intersecciones = [1, 2, 3, 4]
G.add_nodes_from(intersecciones)

# Definimos las calles (Aristas) con pesos dinámicos (Tráfico)
# El peso representa el "costo" en tiempo o densidad vehicular
G.add_edge(1, 2, weight=5)
G.add_edge(2, 3, weight=10) # Calle con mucho tráfico
G.add_edge(3, 4, weight=2)
G.add_edge(4, 1, weight=4)
G.add_edge(2, 4, weight=1) # Atajo detectado por sensor

# 2. Cálculo de la ruta óptima para semáforos inteligentes
inicio = 1
destino = 3
ruta_optima = nx.shortest_path(G, source=inicio, target=destino, weight='weight')

print(f"Ruta óptima calculada para minimizar tiempo: {ruta_optima}")

# 3. Visualización
pos = nx.spring_layout(G)
plt.figure(figsize=(8, 6))

# Dibujar nodos y etiquetas
nx.draw(G, pos, with_labels=True, node_color='skyblue', node_size=1500, font_size=12, font_weight='bold', arrowsize=20)

# Dibujar pesos de las aristas
labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=labels)

plt.title("Simulación de Red Vial - Semáforos Inteligentes")
plt.show()
```
